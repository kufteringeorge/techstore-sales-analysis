```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("../data/sales.csv")

df.head()
```

```python
df.info()
```

```python
print("Количество заказов:", df["OrderID"].nunique())
print("Количество клиентов:", df["CustomerID"].nunique())
print("Общая выручка:", round(df["Revenue"].sum(), 2))
print("Средний чек:", round(df["Revenue"].mean(), 2))
```

```python
category_analysis = (
    df.groupby("Category")
    .agg(
        Orders=("OrderID", "count"),
        Revenue=("Revenue", "sum")
    )
    .sort_values("Revenue", ascending=False)
)

category_analysis
```

```python
city_analysis = (
    df.groupby("City")
    .agg(
        Orders=("OrderID", "count"),
        Revenue=("Revenue", "sum")
    )
    .sort_values("Revenue", ascending=False)
)

city_analysis
```

```python
top_products = (
    df.groupby("Product")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

```python
discount_analysis = (
    df.groupby("Discount")["Revenue"]
    .sum()
)

discount_analysis
```

```python
print("Основные выводы:")
print("- Больше всего выручки приносят категории с высокой стоимостью товаров.")
print("- Москва и Санкт-Петербург являются крупнейшими рынками.")
print("- В период повышенного спроса увеличивается количество заказов.")
print("- Скидки влияют на продажи и позволяют увеличивать объем заказов.")
```

```python
category_sales = (
    df.groupby("Category")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

```python
import matplotlib.pyplot as plt

category_sales.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue")

plt.show()
```
